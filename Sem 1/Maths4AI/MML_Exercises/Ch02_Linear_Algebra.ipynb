## Exercise 2.1 — Abelian Group $(\mathbb{R}\setminus\{-1\}, \star)$

We consider $(\mathbb{R}\setminus\{-1\}, \star)$ with $a \star b := ab + a + b$.

**(a)** Show that $(\mathbb{R}\setminus\{-1\}, \star)$ is an Abelian group.

**(b)** Solve $3 \star x \star x = 15$.

### Solution 2.1

**Key observation:** $a \star b = ab + a + b = (1+a)(1+b) - 1$.

So if we let $\phi(a) = 1 + a$, then $\phi(a \star b) = (1+a)(1+b) = \phi(a) \cdot \phi(b)$.

This is an isomorphism to $(\mathbb{R}\setminus\{0\}, \cdot)$, which is a known Abelian group.

**(a)** We verify all group axioms:

1. **Closure:** For $a,b \in \mathbb{R}\setminus\{-1\}$: $a \star b = (1+a)(1+b) - 1$. Since $1+a \neq 0$ and $1+b \neq 0$, we have $(1+a)(1+b) \neq 0$, so $a \star b \neq -1$. ✓

2. **Associativity:** $(a \star b) \star c = ((1+a)(1+b)-1) \star c = (1+a)(1+b)(1+c) - 1 = a \star (b \star c)$. ✓

3. **Identity:** $e = 0$ since $a \star 0 = a \cdot 0 + a + 0 = a$. ✓

4. **Inverse:** $a^{-1} = \frac{-a}{1+a}$ since $a \star \frac{-a}{1+a} = (1+a)\frac{1}{1+a} - 1 = 0 = e$. ✓

5. **Commutativity:** $a \star b = ab + a + b = ba + b + a = b \star a$. ✓

**(b)** Using the isomorphism: $3 \star x \star x = 15 \implies (1+3)(1+x)^2 - 1 = 15 \implies 4(1+x)^2 = 16 \implies (1+x)^2 = 4 \implies x = 1 \text{ or } x = -3$.

```python
# Verification of Exercise 2.1
def star(a, b):
    return a*b + a + b

# (a) Verify group axioms numerically
a, b, c = 2.5, 3.7, -0.5
print("=== Exercise 2.1 ===")
print(f"Associativity: (a⋆b)⋆c = {star(star(a,b), c):.6f}, a⋆(b⋆c) = {star(a, star(b,c)):.6f}")
print(f"Identity (e=0): a⋆0 = {star(a, 0)}, should be {a}")
a_inv = -a / (1 + a)
print(f"Inverse of {a}: {a_inv:.6f}, a⋆a⁻¹ = {star(a, a_inv):.6f} (should be 0)")
print(f"Commutativity: a⋆b = {star(a,b):.6f}, b⋆a = {star(b,a):.6f}")

# (b) Verify solutions
for x in [1, -3]:
    result = star(3, star(x, x))
    print(f"\n3⋆{x}⋆{x} = {result} (should be 15)")
```

## Exercise 2.2 — Congruence Classes $\mathbb{Z}_n$

**(a)** Show that $(\mathbb{Z}_n, \oplus)$ is a group. Is it Abelian?

**(b)** For $n=5$: Draw the multiplication table of $\mathbb{Z}_5\setminus\{\bar{0}\}$ under $\otimes$. Show it's an Abelian group.

**(c)** Show $(\mathbb{Z}_8\setminus\{\bar{0}\}, \otimes)$ is NOT a group.

**(d)** $(\mathbb{Z}_n\setminus\{\bar{0}\}, \otimes)$ is a group iff $n$ is prime (Bézout's theorem).

### Solution 2.2

**(a)** $(\mathbb{Z}_n, \oplus)$ is a group:
- **Closure:** $(\bar{a} \oplus \bar{b}) = \overline{a+b} \in \mathbb{Z}_n$ ✓
- **Associativity:** Inherited from $(\mathbb{Z}, +)$ ✓
- **Identity:** $\bar{0}$ since $\bar{a} \oplus \bar{0} = \overline{a+0} = \bar{a}$ ✓
- **Inverse:** $\overline{n-a}$ since $\bar{a} \oplus \overline{n-a} = \overline{a + n - a} = \bar{n} = \bar{0}$ ✓
- **Abelian:** $\bar{a} \oplus \bar{b} = \overline{a+b} = \overline{b+a} = \bar{b} \oplus \bar{a}$ ✓

**(c)** In $\mathbb{Z}_8$: $\bar{2} \otimes \bar{4} = \overline{8} = \bar{0}$, so $\bar{2}$ has no inverse (it's a zero divisor). Not a group.

**(d)** By Bézout: $\gcd(a,n)=1 \iff \exists u,v: au + nv = 1 \implies \bar{a}\otimes\bar{u} = \bar{1}$. If $n$ is prime, every $a \in \{1,...,n-1\}$ satisfies $\gcd(a,n)=1$.

```python
# (b) Multiplication table for Z5\{0}
print("=== Exercise 2.2(b): Multiplication table for Z₅\\{0} ===")
n = 5
elements = list(range(1, n))
print(f"\n  ⊗ | {' '.join(str(e) for e in elements)}")
print(f"----+{'---' * len(elements)}")
for a in elements:
    row = [(a * b) % n for b in elements]
    print(f"  {a} | {' '.join(str(r) for r in row)}")

# Verify closure (no zeros in table)
table = [[(a*b)%n for b in elements] for a in elements]
print(f"\nClosure: 0 appears in table? {any(0 in row for row in table)} (should be False)")

# Find inverses
print("\nInverses:")
for a in elements:
    for b in elements:
        if (a * b) % n == 1:
            print(f"  {a}⁻¹ = {b}")
            break

# (c) Z8\{0} is NOT a group
print("\n=== Exercise 2.2(c): Z₈\\{0} ===")
print(f"2 ⊗ 4 = {(2*4) % 8} (mod 8) → zero divisor! Not a group.")

# (d) Check for several n
print("\n=== Exercise 2.2(d): Group iff n is prime ===")
from math import gcd
for n in range(2, 13):
    is_group = all(gcd(a, n) == 1 for a in range(1, n))
    is_prime = all(n % i != 0 for i in range(2, n))
    print(f"  n={n:2d}: is_prime={is_prime}, (Z_n\\{{0}}, ⊗) is group={is_group} {'✓' if is_group == is_prime else '✗'}")
```

## Exercise 2.3 — Heisenberg Group

Is $\left(G = \left\{\begin{bmatrix} 1 & x & z \\ 0 & 1 & y \\ 0 & 0 & 1 \end{bmatrix} : x,y,z \in \mathbb{R}\right\}, \cdot\right)$ a group? Abelian?

### Solution 2.3

Let $A(x_1,y_1,z_1)$ and $A(x_2,y_2,z_2)$ be elements of $G$.

**Closure:**
$$A(x_1,y_1,z_1) \cdot A(x_2,y_2,z_2) = \begin{bmatrix} 1 & x_1+x_2 & z_1+z_2+x_1 y_2 \\ 0 & 1 & y_1+y_2 \\ 0 & 0 & 1 \end{bmatrix} \in G \quad ✓$$

**Associativity:** Matrix multiplication is associative. ✓

**Identity:** $A(0,0,0) = I_3$ ✓

**Inverse:** $A(x,y,z)^{-1} = A(-x, -y, xy - z)$ ✓

**Not Abelian:** $A(1,0,0)\cdot A(0,1,0) = A(1,1,1)$ but $A(0,1,0)\cdot A(1,0,0) = A(1,1,0)$. ✗

$(G, \cdot)$ is a **non-Abelian group** (the Heisenberg group).

```python
# Verification of Exercise 2.3
def heisenberg(x, y, z):
    return np.array([[1, x, z], [0, 1, y], [0, 0, 1]])

A1 = heisenberg(2, 3, 5)
A2 = heisenberg(-1, 4, 7)

print("=== Exercise 2.3 ===")
# Closure
prod = A1 @ A2
print(f"Product A1·A2:\n{prod}")
print(f"Lower-triangular with 1s on diagonal? {np.allclose(prod[1,0], 0) and np.allclose(prod[2,:2], [0,0]) and np.allclose(np.diag(prod), [1,1,1])}")

# Inverse
x, y, z = 2, 3, 5
A_inv = heisenberg(-x, -y, x*y - z)
print(f"\nA·A⁻¹:\n{A1 @ A_inv}")

# Non-commutativity
B1 = heisenberg(1, 0, 0)
B2 = heisenberg(0, 1, 0)
print(f"\nA(1,0,0)·A(0,1,0) =\n{B1 @ B2}")
print(f"A(0,1,0)·A(1,0,0) =\n{B2 @ B1}")
print(f"\nCommutative? {np.allclose(B1 @ B2, B2 @ B1)} → NOT Abelian")
```

## Exercise 2.7 — Eigenvalue Equation & Constraint

Find all $x \in \mathbb{R}^3$ such that $Ax = 12x$ and $\sum_{i} x_i = 1$.

```python
print("=== Exercise 2.7 ===")
A7 = sp.Matrix([[6, 4, 3], [6, 0, 9], [0, 8, 0]])
print("A =")
sp.pprint(A7)

# Ax = 12x  →  (A - 12I)x = 0
A7_shifted = A7 - 12 * sp.eye(3)
print("\nA - 12I =")
sp.pprint(A7_shifted)

# Solve (A - 12I)x = 0
null = A7_shifted.nullspace()
print(f"\nNullspace of (A-12I): {null}")

if null:
    # The eigenvector is proportional to null[0]
    v = null[0]
    # Apply constraint sum(xi) = 1
    s = sum(v)
    x_sol = v / s
    print(f"\nEigenvector (normalized so sum=1):")
    sp.pprint(x_sol)
    
    # Verify
    print(f"\nVerification: Ax = {A7 * x_sol}")
    print(f"12x = {12 * x_sol}")
    print(f"sum(x) = {sum(x_sol)}")
else:
    print("12 is not an eigenvalue of A!")
    print("Eigenvalues of A:", A7.eigenvals())
```

## Exercise 2.9 — Subspaces of $\mathbb{R}^3$

Which of the following sets are subspaces of $\mathbb{R}^3$?

### Solution 2.9

**(a)** $A = \{(\lambda, \lambda + \mu^3, \lambda - \mu^3) \mid \lambda, \mu \in \mathbb{R}\}$

**Not a subspace.** Closure under addition fails. Take $\lambda_1=0, \mu_1=1$ giving $(0, 1, -1)$ and $\lambda_2=0, \mu_2=1$ giving $(0, 1, -1)$. Sum is $(0, 2, -2)$. For this to be in $A$: $\lambda = 0$, $\mu^3 = 2 \implies \mu = \sqrt[3]{2}$, and $-\mu^3 = -2$. ✓ this works. But take $\lambda=0, \mu=1$: $(0,1,-1)$ and $\lambda=0, \mu=2$: $(0,8,-8)$. Sum: $(0,9,-9)$. Need $\mu^3=9$, $\lambda=0$ ✓. Actually, let's check scalar multiplication: $2 \cdot (0,1,-1) = (0,2,-2)$, need $\lambda=0, \mu^3=2$. ✓

Actually, $A = \{(\lambda, \lambda+t, \lambda-t) \mid \lambda \in \mathbb{R}, t \in \mathbb{R}\}$ where $t = \mu^3$ ranges over all of $\mathbb{R}$. So $A = \text{span}\{(1,1,1), (0,1,-1)\}$, which **is a subspace**. ✓

**(b)** $B = \{(\lambda^2, -\lambda^2, 0) \mid \lambda \in \mathbb{R}\}$

**Not a subspace.** $(1,-1,0) \in B$ (take $\lambda=1$), but $(-1)(1,-1,0) = (-1,1,0)$. For $(-1,1,0) \in B$: need $\lambda^2 = -1$, impossible. ✗

**(c)** $C = \{(\xi_1, \xi_2, \xi_3) \in \mathbb{R}^3 \mid \xi_1 - 2\xi_2 + 3\xi_3 = \gamma\}$

- If $\gamma = 0$: **Subspace** (solution set of homogeneous equation). ✓
- If $\gamma \neq 0$: **Not a subspace** ($0 \notin C$). ✗

**(d)** $D = \{(\xi_1, \xi_2, \xi_3) \in \mathbb{R}^3 \mid \xi_2 \in \mathbb{Z}\}$

**Not a subspace.** $(0,1,0) \in D$ but $\frac{1}{2}(0,1,0) = (0, 0.5, 0) \notin D$. ✗

```python
print("=== Exercise 2.9 — Summary ===")
print("(a) A: YES — subspace (span of (1,1,1) and (0,1,-1))")
print("(b) B: NO  — not closed under scalar multiplication")
print("(c) C: YES iff γ=0, NO otherwise")
print("(d) D: NO  — not closed under scalar multiplication")
```

## Exercise 2.12 — Basis of Intersection $U_1 \cap U_2$

```python
print("=== Exercise 2.12 ===")
# U1 = span of columns of A1, U2 = span of columns of A2
A1 = sp.Matrix([[1, 2, -1], [1, -1, 1], [-3, 0, -1], [1, -1, 0]])
A2 = sp.Matrix([[-1, 2, -3], [-2, -2, 6], [2, 0, -2], [1, 1, -1]])

print("U1 generators (columns of A1):")
sp.pprint(A1)
print(f"dim(U1) = rank(A1) = {A1.rank()}")

print("\nU2 generators (columns of A2):")
sp.pprint(A2)
print(f"dim(U2) = rank(A2) = {A2.rank()}")

# U1 ∩ U2: vectors that are in both spans
# x ∈ U1 ∩ U2  ⟺  A1·α = A2·β  ⟺  [A1 | -A2]·[α; β] = 0
M = A1.row_join(-A2)
print("\n[A1 | -A2]:")
sp.pprint(M)

ns = M.nullspace()
print(f"\nNullspace dimension: {len(ns)}")

# Extract intersection basis: for each nullspace vector [α; β], compute A1·α
intersection_basis = []
for v in ns:
    alpha = v[:A1.cols, :]
    vec = A1 * alpha
    intersection_basis.append(vec)

print("\nBasis for U1 ∩ U2:")
for v in intersection_basis:
    sp.pprint(v.T)
```

## Exercise 2.13 — Solution Spaces Intersection

## Exercise 2.15 — Subspace Intersection $F \cap G$

```python
print("=== Exercise 2.15 ===")
print("F = {(x,y,z) ∈ ℝ³ | x + y - z = 0}")
print("G = {(a-b, a+b, a-3b) | a,b ∈ ℝ}")

# (a) Show F and G are subspaces
print("\n(a) F is the kernel of [1, 1, -1] → subspace ✓")
print("    G = span{(1,1,1), (-1,1,-3)} → subspace ✓")

# (b) F ∩ G without basis vectors
# (x,y,z) ∈ G means x=a-b, y=a+b, z=a-3b
# (x,y,z) ∈ F means x+y-z = 0
# Substituting: (a-b)+(a+b)-(a-3b) = a+3b = 0 → a = -3b
# So x = -3b-b = -4b, y = -3b+b = -2b, z = -3b-3b = -6b
print("\n(b) Substituting G into F's equation:")
print("    (a-b)+(a+b)-(a-3b) = a + 3b = 0 → a = -3b")
print("    F ∩ G = {(-4b, -2b, -6b) | b ∈ ℝ} = span{(-4, -2, -6)} = span{(2, 1, 3)}")

# (c) Using basis vectors
print("\n(c) Verification using basis computation:")
# F: kernel of [1, 1, -1]
F_eq = sp.Matrix([[1, 1, -1]])
F_basis = F_eq.nullspace()
print("Basis of F:")
for v in F_basis:
    sp.pprint(v.T)

# G = span{(1,1,1), (-1,1,-3)}
G_gens = sp.Matrix([[1, -1], [1, 1], [1, -3]])
print("Generators of G: (1,1,1) and (-1,1,-3)")

# F ∩ G: solve [G_gens | -F_basis_matrix] · params = 0
F_mat = sp.Matrix.hstack(*F_basis)  # columns
M_15 = G_gens.row_join(-F_mat)
ns_15 = M_15.nullspace()
print("\nBasis of F ∩ G:")
for v in ns_15:
    vec = G_gens * v[:2, :]
    sp.pprint(vec.T)
```

## Exercise 2.16 — Linear Mappings

Are the following mappings linear?

### Solution 2.16

**(a)** $\Phi(f) = \int_a^b f(x)\,dx$ — **Linear** ✓
- $\Phi(f+g) = \int(f+g) = \int f + \int g = \Phi(f) + \Phi(g)$
- $\Phi(\alpha f) = \int \alpha f = \alpha \int f = \alpha\Phi(f)$

**(b)** $\Phi(f) = f'$ (differentiation) — **Linear** ✓
- $(f+g)' = f' + g'$
- $(\alpha f)' = \alpha f'$

**(c)** $\Phi(x) = \cos(x)$ — **Not linear** ✗
- $\cos(x+y) \neq \cos(x) + \cos(y)$ in general
- Counter-example: $\cos(0) = 1 \neq 0$, but $\Phi(0) = 0$ is required for linear maps.

**(d)** $\Phi(x) = \begin{pmatrix} 1 & 2 & 3 \\ 1 & 4 & 3 \end{pmatrix} x$ — **Linear** ✓
- Matrix-vector multiplication is always linear.

**(e)** $\Phi(x) = \begin{pmatrix} \cos\theta & \sin\theta \\ -\sin\theta & \cos\theta \end{pmatrix} x$ — **Linear** ✓
- Matrix multiplication = linear mapping (rotation matrix).

```python
print("=== Exercise 2.16 — Summary ===")
print("(a) Integration:      LINEAR  ✓")
print("(b) Differentiation:  LINEAR  ✓")
print("(c) cos(x):           NOT linear ✗ (cos(0) = 1 ≠ 0)")
print("(d) Matrix multiply:  LINEAR  ✓")
print("(e) Rotation matrix:  LINEAR  ✓")

# Verify (c)
print(f"\nVerification (c): cos(0) = {np.cos(0)} ≠ 0, so Φ(0) ≠ 0 → not linear")
print(f"Also: cos(π/4 + π/4) = {np.cos(np.pi/4 + np.pi/4):.4f}")
print(f"  but cos(π/4) + cos(π/4) = {np.cos(np.pi/4) + np.cos(np.pi/4):.4f}")
```

## Exercise 2.18 — Kernel and Image of Composed Automorphisms

Let $f \circ g = \text{id}_E$. Show:
1. $\ker(f) = \ker(g \circ f)$
2. $\text{Im}(g) = \text{Im}(g \circ f)$
3. $\ker(f) \cap \text{Im}(g) = \{0_E\}$

### Solution 2.18

Given: $f \circ g = \text{id}_E$.

**1. $\ker(f) = \ker(g \circ f)$:**

- ($\subseteq$): If $x \in \ker(f)$, then $f(x) = 0$, so $(g \circ f)(x) = g(f(x)) = g(0) = 0$, hence $x \in \ker(g \circ f)$.
- ($\supseteq$): If $x \in \ker(g \circ f)$, then $g(f(x)) = 0$. Apply $f$: $f(g(f(x))) = f(0) = 0$. But $f \circ g = \text{id}$, so $f(g(f(x))) = f(x)$. Thus $f(x) = 0$, i.e., $x \in \ker(f)$.

**2. $\text{Im}(g) = \text{Im}(g \circ f)$:**

- ($\supseteq$): $\text{Im}(g \circ f) \subseteq \text{Im}(g)$ trivially (since $g \circ f(x) = g(f(x)) \in \text{Im}(g)$).
- ($\subseteq$): Let $y \in \text{Im}(g)$. Then $y = g(z)$ for some $z$. Since $f \circ g = \text{id}$, we have $f(g(z)) = z$. So $y = g(z) = g(f(g(z))) = (g \circ f)(g(z)) \in \text{Im}(g \circ f)$.

**3. $\ker(f) \cap \text{Im}(g) = \{0_E\}$:**

Let $x \in \ker(f) \cap \text{Im}(g)$. Then $f(x) = 0$ and $x = g(z)$ for some $z$.
So $0 = f(x) = f(g(z)) = z$ (since $f \circ g = \text{id}$). Thus $x = g(0) = 0$. ∎

```python
# Numerical verification with concrete matrices
print("=== Exercise 2.18 — Numerical Verification ===")
# f ∘ g = id, so f = g⁻¹ on Im(g)
# Let's use f as a 3×3 matrix F and g = right-inverse
# For square invertible matrices: F·G = I means G = F⁻¹
F = sp.Matrix([[1, 2, 0], [0, 1, 1], [1, 0, 1]])
G = F.inv()
print(f"F·G = I: {F * G == sp.eye(3)}")
print(f"ker(F) = {F.nullspace()} (trivial for invertible)")
print(f"ker(G∘F) = {(G*F).nullspace()} (G·F = I, so trivial)")
print("Properties verified for invertible case (all trivial).")

# For non-square case (more interesting)
print("\nNon-square case: E = ℝ³, F: ℝ³→ℝ², G: ℝ²→ℝ³")
F2 = sp.Matrix([[1, 0, 0], [0, 1, 0]])  # projection
G2 = sp.Matrix([[1, 0], [0, 1], [0, 0]])  # embedding
print(f"F·G = {F2 * G2} = I₂ ✓")
print(f"ker(F) = {[v.T for v in F2.nullspace()]}")
print(f"ker(G∘F) = {[v.T for v in (G2*F2).nullspace()]}")
print(f"Im(G) = {[v.T for v in G2.columnspace()]}")
print(f"Im(G∘F) = {[v.T for v in (G2*F2).columnspace()]}")
```

## Exercise 2.19 — Kernel, Image, and Basis Change

```python
print("=== Exercise 2.19 ===")
A_19 = sp.Matrix([[1, 1, 0], [1, -1, 0], [1, 1, 1]])
print("A_Φ (standard basis):")
sp.pprint(A_19)

# (a) ker(Φ) and Im(Φ)
print("\n(a) ker(Φ):")
ker_19 = A_19.nullspace()
if ker_19:
    for v in ker_19:
        sp.pprint(v.T)
else:
    print("  ker(Φ) = {0}")

print("\nIm(Φ):")
for v in A_19.columnspace():
    sp.pprint(v.T)
print(f"dim(Im(Φ)) = {len(A_19.columnspace())}")

# (b) Basis change to B = {b1, b2, b3}
print("\n(b) Basis change:")
B = sp.Matrix([[1, 1, 1], [1, 2, 0], [1, 1, 0]])
print("B (new basis as columns):")
sp.pprint(B)

# Ã = B⁻¹ · A · B
A_tilde = B.inv() * A_19 * B
print("\nÃ_Φ = B⁻¹ · A_Φ · B:")
sp.pprint(A_tilde)

# Verify: A_Φ · b_i should map to same as B · Ã · e_i
print("\nVerification:")
for i in range(3):
    e_i = sp.zeros(3, 1)
    e_i[i] = 1
    lhs = A_19 * B[:, i]
    rhs = B * A_tilde * e_i
    print(f"  A·b{i+1} = {lhs.T},  B·Ã·e{i+1} = {rhs.T}, match: {lhs == rhs}")
```

## Exercise 2.20 — Basis Change in $\mathbb{R}^2$

```python
print("=== Exercise 2.20 ===")
# b1, b2 expressed in standard basis
b1 = sp.Matrix([2, 1])
b2 = sp.Matrix([-1, -1])

# b1', b2' expressed in standard basis  
b1p = sp.Matrix([2, -2])
b2p = sp.Matrix([1, 1])

# B = [b1 | b2], B' = [b1' | b2']
B_old = sp.Matrix([[2, -1], [1, -1]])
B_new = sp.Matrix([[2, 1], [-2, 1]])

print("Old basis B:")
sp.pprint(B_old)
print("\nNew basis B':")
sp.pprint(B_new)

# Linear mapping Φ given in B coordinates:
# Φ(b1) = b1 - b2 and Φ(b2) = b1 + b2 (in old basis)
# So the matrix w.r.t. B is:
A_B = sp.Matrix([[1, 1], [-1, 1]])  # columns = images of basis vectors
print("\nA_Φ w.r.t. old basis B:")
sp.pprint(A_B)

# Transformation matrix w.r.t. standard basis:
A_std = B_old * A_B * B_old.inv()
print("\nA_Φ w.r.t. standard basis:")
sp.pprint(A_std)

# Transformation matrix w.r.t. new basis B':
A_Bp = B_new.inv() * A_std * B_new
# Or equivalently: A_Bp = B_new.inv() * B_old * A_B * B_old.inv() * B_new
print("\nA_Φ w.r.t. new basis B':")
sp.pprint(A_Bp)

# The change of basis matrix S from B to B'
S = B_new.inv() * B_old
print("\nChange of basis matrix S = (B')⁻¹ B:")
sp.pprint(S)
print(f"\nVerification: S⁻¹ · A_B · S = A_B'?")
sp.pprint(S.inv() * A_B * S)
```

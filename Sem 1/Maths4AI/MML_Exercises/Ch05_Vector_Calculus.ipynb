## Exercise 5.1 — Derivative of $f(x) = \log(x^4) \sin(x^3)$

```python
print("=== Exercise 5.1 ===")
x = sp.Symbol('x', positive=True)
f51 = log(x**4) * sin(x**3)
f51_prime = diff(f51, x)

print("f(x) ="); pprint(f51)
print("\nf'(x) ="); pprint(f51_prime)
print("\nSimplified:"); pprint(simplify(f51_prime))

# Using product rule manually: (uv)' = u'v + uv'
print("\n--- Manual derivation ---")
print("f = log(x⁴)·sin(x³) = 4·log(x)·sin(x³)")
print("f' = (4/x)·sin(x³) + 4·log(x)·3x²·cos(x³)")
print("   = 4sin(x³)/x + 12x²·log(x)·cos(x³)")
```

## Exercise 5.2 — Derivative of Logistic Sigmoid

```python
print("=== Exercise 5.2 ===")
x = sp.Symbol('x')
sigma = 1 / (1 + exp(-x))
sigma_prime = diff(sigma, x)

print("σ(x) ="); pprint(sigma)
print("\nσ'(x) ="); pprint(simplify(sigma_prime))

# Show the famous identity σ'(x) = σ(x)(1 - σ(x))
identity = simplify(sigma * (1 - sigma) - sigma_prime)
print(f"\nσ'(x) = σ(x)·(1-σ(x))? Difference = {identity} ✓")

# Plot
x_vals = np.linspace(-6, 6, 200)
sig_vals = 1 / (1 + np.exp(-x_vals))
sig_prime_vals = sig_vals * (1 - sig_vals)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x_vals, sig_vals, 'b-', lw=2, label=r'$\sigma(x)$')
ax.plot(x_vals, sig_prime_vals, 'r--', lw=2, label=r"$\sigma'(x) = \sigma(1-\sigma)$")
ax.legend(fontsize=12); ax.grid(alpha=0.3); ax.set_xlabel('x')
ax.set_title('Exercise 5.2: Logistic Sigmoid and its Derivative')
plt.tight_layout(); plt.show()
```

## Exercise 5.3 — Derivative of Gaussian Kernel

```python
print("=== Exercise 5.3 ===")
x, mu, sigma_s = sp.symbols('x mu sigma', real=True)
f53 = exp(-1 / (2 * sigma_s**2) * (x - mu)**2)
f53_prime = diff(f53, x)

print("f(x) ="); pprint(f53)
print("\nf'(x) ="); pprint(simplify(f53_prime))
print("\n= -(x-μ)/σ² · exp(-(x-μ)²/(2σ²))")
```

## Exercise 5.4 — Taylor Polynomials of $\sin(x) + \cos(x)$

```python
print("=== Exercise 5.4 ===")
x = sp.Symbol('x')
f54 = sin(x) + cos(x)

print("f(x) = sin(x) + cos(x)")
print("\nTaylor polynomials at x₀ = 0:")
for n in range(6):
    Tn = sp.series(f54, x, 0, n+1).removeO()
    print(f"  T{n}(x) = {Tn}")

# Plot
x_vals = np.linspace(-3, 3, 200)
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x_vals, np.sin(x_vals) + np.cos(x_vals), 'k-', lw=3, label='f(x)', alpha=0.7)
colors = plt.cm.viridis(np.linspace(0.2, 0.9, 6))
for n in range(6):
    Tn = sp.series(f54, x, 0, n+1).removeO()
    Tn_func = sp.lambdify(x, Tn, 'numpy')
    try:
        y_vals = Tn_func(x_vals)
        if np.isscalar(y_vals):
            y_vals = np.full_like(x_vals, y_vals)
        ax.plot(x_vals, y_vals, '--', color=colors[n], lw=1.5, label=f'T{n}')
    except:
        pass
ax.set_ylim(-3, 3); ax.legend(fontsize=9); ax.grid(alpha=0.3)
ax.set_title('Exercise 5.4: Taylor Polynomials of sin(x) + cos(x)')
plt.tight_layout(); plt.show()
```

## Exercise 5.5 — Jacobians of Multivariate Functions

```python
print("=== Exercise 5.5 ===")

# f1(x) = sin(x1)cos(x2), x ∈ R²
print("--- f₁(x) = sin(x₁)cos(x₂) ---")
x1, x2 = sp.symbols('x1 x2')
f1 = sin(x1) * cos(x2)
J1 = sp.Matrix([f1]).jacobian([x1, x2])
print("∂f₁/∂x is 1×2 (scalar function, 2D input)")
print("Jacobian:"); pprint(J1)

# f2(x,y) = x^T y, x,y ∈ Rⁿ
print("\n--- f₂(x,y) = xᵀy ---")
print("∂f₂/∂x = yᵀ  (1×n)")
print("∂f₂/∂y = xᵀ  (1×n)")
# Verify with n=3
x_vec = sp.Matrix(sp.symbols('x1:4'))
y_vec = sp.Matrix(sp.symbols('y1:4'))
f2 = (x_vec.T * y_vec)[0,0]
J2x = sp.Matrix([f2]).jacobian(x_vec)
J2y = sp.Matrix([f2]).jacobian(y_vec)
print(f"∂(xᵀy)/∂x = {J2x} = yᵀ ✓")
print(f"∂(xᵀy)/∂y = {J2y} = xᵀ ✓")

# f3(x) = xxᵀ, x ∈ Rⁿ → output is n×n
print("\n--- f₃(x) = xxᵀ ---")
print("∂f₃/∂x: f₃ maps Rⁿ → R^(n×n), so Jacobian is n²×n")
print("Each entry: ∂(xᵢxⱼ)/∂xₖ = δᵢₖ·xⱼ + xᵢ·δⱼₖ")
# Verify with n=2
x_2 = sp.Matrix(sp.symbols('x1 x2'))
f3 = x_2 * x_2.T  # 2×2 matrix
print("For n=2, xxᵀ ="); pprint(f3)
# Flatten and compute Jacobian
f3_flat = sp.Matrix([f3[i,j] for i in range(2) for j in range(2)])
J3 = f3_flat.jacobian(x_2)
print("Jacobian (4×2):")
pprint(J3)
```

## Exercise 5.6 — Derivatives of Trace and Sin-Log Functions

```python
print("=== Exercise 5.6 ===")

# f(t) = sin(log(tᵀt)), t ∈ R^D
print("--- f(t) = sin(log(tᵀt)) ---")
print("Chain rule: df/dt = cos(log(tᵀt)) · 1/(tᵀt) · 2tᵀ")
print("          = 2cos(log(tᵀt))/(tᵀt) · tᵀ")

# Verify with D=3
t = sp.Matrix(sp.symbols('t1 t2 t3'))
f_t = sin(log(t.dot(t)))
grad_t = sp.Matrix([diff(f_t, ti) for ti in t]).T
print("\nSymPy gradient (1×3):")
pprint(simplify(grad_t))

# g(X) = tr(AXB)
print("\n--- g(X) = tr(AXB) ---")
print("dg/dX = Aᵀ · (dtr/d(AXB)) · Bᵀ")
print("Since d(tr(M))/dM = I, and using the identity:")
print("dg/dXᵢⱼ = (Aᵀ)ᵢ · (Bᵀ)ⱼ")
print("So: dg/dX = AᵀBᵀ = (BA)ᵀ")

# Verify with small matrices
print("\nVerification with 2×2 case:")
a11, a12, a21, a22 = sp.symbols('a11 a12 a21 a22')
x11, x12, x21, x22 = sp.symbols('x11 x12 x21 x22')
b11, b12, b21, b22 = sp.symbols('b11 b12 b21 b22')
A_mat = sp.Matrix([[a11, a12], [a21, a22]])
X_mat = sp.Matrix([[x11, x12], [x21, x22]])
B_mat = sp.Matrix([[b11, b12], [b21, b22]])
g = (A_mat * X_mat * B_mat).trace()
dg_dX = sp.Matrix([[diff(g, x11), diff(g, x12)], [diff(g, x21), diff(g, x22)]])
expected = (B_mat * A_mat).T  # = A^T B^T
print(f"dg/dX = A^T B^T? {simplify(dg_dX - expected) == sp.zeros(2)} ✓")
```

## Exercise 5.7 — Chain Rule Derivatives

```python
print("=== Exercise 5.7 ===")

# (a) f(z) = log(1+z), z = xᵀx, x ∈ R^D
print("--- (a) f(z) = log(1+z), z = xᵀx ---")
print("df/dz = 1/(1+z)    (scalar → scalar: 1×1)")
print("dz/dx = 2xᵀ        (scalar → R^D: 1×D)")
print("df/dx = df/dz · dz/dx = 2xᵀ/(1+xᵀx)  (1×D)")

# Verify with D=3
x = sp.Matrix(sp.symbols('x1 x2 x3'))
z_a = x.dot(x)
f_a = log(1 + z_a)
grad_a = sp.Matrix([diff(f_a, xi) for xi in x]).T
print("SymPy:")
pprint(simplify(grad_a))

# (b) f(z) = sin(z), z = Ax + b
print("\n--- (b) f(z) = sin(z), z = Ax + b ---")
print("df/dz = diag(cos(z))   (E×E, elementwise)")
print("dz/dx = A               (E×D)")
print("df/dx = diag(cos(Ax+b)) · A   (E×D)")
```

## Exercise 5.8 — Chain Rule with Gaussian and Trace

```python
print("=== Exercise 5.8 ===")

# (a) f(z) = exp(-z/2), z = yᵀS⁻¹y, y = x - μ
print("--- (a) Chain rule for Gaussian exponent ---")
print("df/dz = -½ exp(-z/2)                    (1×1)")
print("dz/dy = 2yᵀS⁻¹ = (S⁻¹ + S⁻ᵀ)y = 2S⁻¹y  (since S symmetric)  (1×D)")
print("dy/dx = I_D                              (D×D)")
print("")
print("df/dx = df/dz · dz/dy · dy/dx")
print("      = -½ exp(-½(x-μ)ᵀS⁻¹(x-μ)) · 2(x-μ)ᵀS⁻¹ · I")
print("      = -(x-μ)ᵀS⁻¹ · exp(-½(x-μ)ᵀS⁻¹(x-μ))    (1×D)")

# (b) f(x) = tr(xxᵀ + σ²I)
print("\n--- (b) f(x) = tr(xxᵀ + σ²I) ---")
print("tr(xxᵀ + σ²I) = tr(xxᵀ) + tr(σ²I) = xᵀx + Dσ²")
print("df/dx = 2xᵀ    (1×D)")

# Verify
x = sp.Matrix(sp.symbols('x1 x2'))
sigma2 = sp.Symbol('sigma2')
f_b = (x * x.T + sigma2 * sp.eye(2)).trace()
print(f"\nVerify: tr(xxᵀ+σ²I) = {f_b}")
grad_b = sp.Matrix([diff(f_b, xi) for xi in x]).T
print(f"df/dx = {grad_b} = 2xᵀ ✓")

# (c) f = tanh(z), z = Ax + b
print("\n--- (c) f = tanh(z), z = Ax + b ---")
print("df/dz = diag(1 - tanh²(z))    (M×M)")
print("dz/dx = A                      (M×N)")
print("df/dx = diag(1 - tanh²(Ax+b)) · A    (M×N)")
```

## Exercise 5.9 — Gradient of ELBO-like Expression

### Solution 5.9

Given:
$$g(z, \nu) = \log p(x, z) - \log q(z, \nu), \quad z = t(\epsilon, \nu)$$

Compute $\frac{d}{d\nu} g(z, \nu)$:

By the chain rule:
$$\frac{dg}{d\nu} = \frac{\partial g}{\partial z} \cdot \frac{\partial z}{\partial \nu} + \frac{\partial g}{\partial \nu}$$

Where:
- $\frac{\partial g}{\partial z} = \frac{\partial \log p(x,z)}{\partial z} - \frac{\partial \log q(z,\nu)}{\partial z}$
- $\frac{\partial z}{\partial \nu} = \frac{\partial t(\epsilon, \nu)}{\partial \nu}$
- $\frac{\partial g}{\partial \nu} = -\frac{\partial \log q(z,\nu)}{\partial \nu}$ (explicit dependence on $\nu$)

Therefore:
$$\frac{dg}{d\nu} = \left(\frac{\partial \log p(x,z)}{\partial z} - \frac{\partial \log q(z,\nu)}{\partial z}\right) \frac{\partial t(\epsilon,\nu)}{\partial \nu} - \frac{\partial \log q(z,\nu)}{\partial \nu}$$

This is the **reparameterization trick** gradient, fundamental to VAEs (Variational Autoencoders).

```python
print("=== Exercise 5.9 ===")
print("This is the reparameterization trick gradient:")
print("")
print("dg/dν = (∂log p(x,z)/∂z - ∂log q(z,ν)/∂z) · ∂t(ε,ν)/∂ν - ∂log q(z,ν)/∂ν")
print("")
print("Concrete example: Gaussian variational posterior")
print("q(z|ν) = N(z; μ, σ²) where ν = (μ, σ)")
print("Reparameterization: z = t(ε, ν) = μ + σε, ε ~ N(0,1)")
print("")
print("∂z/∂μ = 1, ∂z/∂σ = ε")
print("∂log q/∂μ = (z-μ)/σ², ∂log q/∂σ = (z-μ)²/σ³ - 1/σ")
print("")
print("This enables backpropagation through stochastic sampling.")
```

```python
# Problem 3: Statistical Functions

# Data definition
X = [40, 40, 40, 30, 15, 15, 15, 30, 15, 30, 30, 30, 40, 30]
Y = [90, 90, 90, 90, 70, 70, 70, 90, 70, 70, 70, 90, 70, 90]

# 1. Function to calculate Mean
def calculate_mean(data):
    total = sum(data)
    count = len(data)
    return total / count

# 2. Function to calculate Variance (Sample Variance)
def calculate_variance(data):
    mean = calculate_mean(data)
    n = len(data)
    # Variance formula: sum((x - mean)^2) / (n - 1)
    squared_diffs = sum((x - mean) ** 2 for x in data)
    return squared_diffs / (n - 1)

# 3. Function to calculate Standard Deviation
def calculate_stdev(data):
    variance = calculate_variance(data)
    # Square root of variance is standard deviation
    return variance ** 0.5

# 4. Function to calculate Covariance
def calculate_covariance(data_x, data_y):
    if len(data_x) != len(data_y):
        return "Error: Lists must be the same length"
    
    n = len(data_x)
    mean_x = calculate_mean(data_x)
    mean_y = calculate_mean(data_y)
    
    # Covariance formula: sum((x - mean_x) * (y - mean_y)) / (n - 1)
    sum_product_diffs = 0
    for i in range(n):
        diff_x = data_x[i] - mean_x
        diff_y = data_y[i] - mean_y
        sum_product_diffs += diff_x * diff_y
        
    return sum_product_diffs / (n - 1)

# --- Testing the Functions ---
print(f"Mean of X: {calculate_mean(X):.2f}")     # Expected: ~28.57
print(f"Mean of Y: {calculate_mean(Y):.2f}")     # Expected: ~80.00

print(f"Variance of X: {calculate_variance(X):.2f}")
print(f"Variance of Y: {calculate_variance(Y):.2f}")

print(f"Std Dev of X: {calculate_stdev(X):.2f}")
print(f"Std Dev of Y: {calculate_stdev(Y):.2f}")

print(f"Covariance(X, Y): {calculate_covariance(X, Y):.2f}")
```

```python
# Part a

# Data
scores = [60, 80, 90, 50, 65, 75, 78, 82]

def operate_on_scores(data, operation):
    """Applies a lambda operation to the provided data list."""
    return operation(data)

# 1. Return max and min score as a tuple
max_min = operate_on_scores(scores, lambda s: (max(s), min(s)))
print(f"Max and Min: {max_min}")

# 2. Calculate the mean score
mean_val = operate_on_scores(scores, lambda s: sum(s) / len(s))
print(f"Mean Score: {mean_val}")

# 3. Filter scores greater or equal to 65
passed = operate_on_scores(scores, lambda s: list(filter(lambda x: x >= 65, s)))
print(f"Scores >= 65: {passed}")

# 4. Map scores to letter grades (A: 70-100, B: 60-69, C: 50-59)
grades = operate_on_scores(scores, lambda s: list(map(
    lambda x: 'A' if x >= 70 else ('B' if x >= 60 else ('C' if x >= 50 else 'Fail')), s
)))
print(f"Letter Grades: {grades}")
```

**Python Classes**

The base Person class, the Student class with validation logic, the polymorphic UnderGraduate and PostGraduate subclasses, and the StudentDataset container class.

```python
class Person:
    """
    a) Base Class Person
    - Tracks population count using a class variable.
    - Provides string representations (__str__, __repr__).
    """
    population_count = 0  # Class variable

    def __init__(self, id_num, name):
        self.id = id_num
        self.name = name
        Person.population_count += 1

    def display_info(self):
        print(f"ID: {self.id}, Name: {self.name}")

    @classmethod
    def get_population(cls):
        """Class method to return the total number of Person objects created."""
        return cls.population_count

    def __str__(self):
        """Human-readable description."""
        return f"Person(Name: {self.name}, ID: {self.id})"

    def __repr__(self):
        """Developer-friendly representation."""
        return f"Person(id={self.id!r}, name={self.name!r})"


class Student(Person):
    """
    b) Student Class (Inherits from Person)
    - Validates scores using a static method.
    - Calculates weighted final score.
    - Implements __eq__ for ID comparison.
    """
    
    def __init__(self, id_num, name, attendance, assignment, exam):
        super().__init__(id_num, name)
        
        # Validate scores before assigning
        if not (self.validate_score(attendance) and 
                self.validate_score(assignment) and 
                self.validate_score(exam)):
            raise ValueError(f"Error for {name}: All scores must be between 0 and 100.")

        self.evaluation = {
            'attendance': attendance,
            'assignment': assignment,
            'exam': exam
        }

    @staticmethod
    def validate_score(score):
        """Static method to ensure scores are within range 0-100."""
        return 0 <= score <= 100

    def calculate_final_score(self):
        """
        Calculates score based on weighted average:
        Attendance(10%) + Assignment(30%) + Exam(60%)
        """
        return (self.evaluation['attendance'] * 0.10 +
                self.evaluation['assignment'] * 0.30 +
                self.evaluation['exam'] * 0.60)

    def predict_performance(self):
        """Base method for polymorphism (to be overridden)."""
        return "Prediction not defined for generic student."

    def __eq__(self, other):
        """Compares students based on ID."""
        if isinstance(other, Student):
            return self.id == other.id
        return False

    def __lt__(self, other):
        """
        c) Magic method __lt__ 
        Compares students based on Final Score. 
        """
        if isinstance(other, Student):
            return self.calculate_final_score() < other.calculate_final_score()
        return NotImplemented

    def __str__(self):
        return f"Student: {self.name} (ID: {self.id}), Score: {self.calculate_final_score():.2f}"


class UnderGraduate(Student):
    """c) Subclass for Undergraduate students with specific prediction logic."""
    
    def predict_performance(self):
        score = self.calculate_final_score()
        if score >= 70:
            return "Excellent - On track for First Class Honors."
        elif score >= 50:
            return "Good - Likely to pass."
        else:
            return "At Risk - Needs immediate academic counseling."


class PostGraduate(Student):
    """c) Subclass for Postgraduate students with specific prediction logic."""
    
    def predict_performance(self):
        score = self.calculate_final_score()
        if score >= 80:
            return "Outstanding - Potential for research publication."
        elif score >= 60:
            return "Satisfactory - Meets graduate requirements."
        else:
            return "Critical - Performance below graduate standards."


class StudentDataset:
    """
    d) StudentDataset Container Class
    - Manages a private list of students.
    - Implements magic methods for container behavior (len, contains, iter).
    """
    
    def __init__(self):
        self._students = []  # Private list

    def add_student(self, student):
        if isinstance(student, Student):
            self._students.append(student)
        else:
            raise TypeError("Only Student objects can be added.")

    def get_best_performing_student(self):
        """Returns the student with the highest final score."""
        if not self._students:
            return None
        # Uses the logic defined in calculate_final_score
        return max(self._students, key=lambda s: s.calculate_final_score())

    def get_students_at_risk(self, threshold=50):
        """Returns a list of students scoring below a threshold."""
        return [s for s in self._students if s.calculate_final_score() < threshold]

    # Magic Methods 
    
    def __len__(self):
        """Returns the number of students."""
        return len(self._students)

    def __contains__(self, student_id):
        """Checks if a student exists by ID."""
        return any(s.id == student_id for s in self._students)

    def __iter__(self):
        """Allows iterating over the dataset."""
        return iter(self._students)

    def __getitem__(self, index):
        """Allows indexing (e.g., dataset[0])."""
        return self._students[index]


# Testing the Application

if __name__ == "__main__":
    # 1. Create Data
    # UG: Attendance 90, Assignment 80, Exam 85 -> Final: 9+24+51 = 84.0
    s1 = UnderGraduate(101, "Alice", 90, 80, 85)
    
    # PG: Attendance 50, Assignment 50, Exam 40 -> Final: 5+15+24 = 44.0
    s2 = PostGraduate(102, "Bob", 50, 50, 40)
    
    # UG: Attendance 70, Assignment 70, Exam 65 -> Final: 7+21+39 = 67.0
    s3 = UnderGraduate(103, "Charlie", 70, 70, 65)

    # 2. Test Person Class Functionality
    print(f"Total Persons Created: {Person.get_population()}")
    print(repr(s1))  # Developer representation

    # 3. Test StudentDataset
    db = StudentDataset()
    db.add_student(s1)
    db.add_student(s2)
    db.add_student(s3)

    print(f"\nDataset Size: {len(db)}")
    
    # 4. Analyze Performance (Polymorphism in action)
    print("\n Performance Predictions")
    for student in db:
        print(f"{student.name}: {student.predict_performance()}")

    # 5. Best Student & Risk Analysis
    best = db.get_best_performing_student()
    print(f"\nTop Student: {best.name} with score {best.calculate_final_score():.2f}")
    
    at_risk = db.get_students_at_risk(threshold=50)
    print(f"Students at Risk (<50): {[s.name for s in at_risk]}")

    # 6. Comparisons (using __lt__ and __eq__)
    print(f"\nComparison: Is Bob's score less than Alice's score? {s2 < s1}")
```

Using python’s numpy package<br>
    i. Load data from CSV file into a NumPy array(Hint: use numpy’s genfromtxt() or ’open’ the file for reading and use csv reader).<br>
    ii. Display the total number of observations(records)<br>
    iii. Calculate and display the average temperature<br>
    iv. Calculate and display the standard deviation<br>
    v. Determine the max and minimum temperatures. Display the min and max temperatures and the time they were recorded.<br>
    vi. Determine and display the number of temperature observations that are above and below the mean<br>
    vii. Extract 10 main deviations from the mean

```python
import numpy as np

# LOAD DATA FROM LOCAL CSV 
dataset = np.genfromtxt('KENTemp.csv', delimiter=',', names=True, dtype=None, encoding='utf-8')

# Create a variable for the temperature column for easier math
temps = dataset['temperature']

# CALCULATE STATISTICS

# ii. Display total number of observations
total_obs = len(temps)
print(f"ii.  Total Observations: {total_obs}")

# iii. Calculate Average Temperature
mean_temp = np.mean(temps)
print(f"iii. Average Temperature: {mean_temp:.2f}")

# iv. Calculate Standard Deviation
std_dev = np.std(temps)
print(f"iv.  Standard Deviation: {std_dev:.2f}")

# v. Determine Max/Min temperatures and the time they were recorded
# argmax() and argmin() return the *index* (row number) of the highest/lowest value.
max_idx = np.argmax(temps)
min_idx = np.argmin(temps)

# We use those indices to grab the corresponding 'Time' from the dataset
print(f"v.   Max Temp: {temps[max_idx]} at {dataset['date'][max_idx]}")
print(f"     Min Temp: {temps[min_idx]} at {dataset['date'][min_idx]}")

# vi. Number of observations above and below mean
# (temps > mean_temp) creates a list of True/False, and sum() counts the Trues.
count_above = np.sum(temps > mean_temp)
count_below = np.sum(temps < mean_temp)

print(f"vi.  Count Above Mean: {count_above}")
print(f"     Count Below Mean: {count_below}")

# vii. Extract 10 main deviations (The 10 values furthest from the average)
# Calculate the absolute difference (|x - mean|) for every data point
deviations = np.abs(temps - mean_temp)

# argsort sorts indices from smallest deviation to largest.
# [::-1] reverses it (largest to smallest).
# [:10] grabs the top 10.
top_deviation_indices = np.argsort(deviations)[::-1][:10]

print("\nvii. Top 10 Main Deviations (Furthest from Mean):")
print(f"{'date':<10} {'Temp':<10} {'Deviation':<10}")
print("-" * 35)

for i in top_deviation_indices:
    # Use the index 'i' to get the specific Time, Temp, and Deviation for that row
    print(f"{dataset['date'][i]:<10} {temps[i]:<10} {deviations[i]:.2f}")
```

7. Visualize sales trends over time using integration with libraries like Matplotlib.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('sales.csv', encoding='ISO-8859-1')
print("Data loaded. Shape:", df.shape)

# Check missing values
print("Missing values before cleaning:\n", df.isnull().sum()[df.isnull().sum() > 0])

# DATA CLEANING - Imputation Strategy
# 1. Handle TERRITORY: Fill based on Country match
# Create a dictionary mapping: {Country: Territory} from rows where Territory exists
country_territory_map = df.dropna(subset=['TERRITORY']).drop_duplicates('COUNTRY').set_index('COUNTRY')['TERRITORY'].to_dict()

# Use this map to fill missing territory values based on the 'COUNTRY' column
df['TERRITORY'] = df['TERRITORY'].fillna(df['COUNTRY'].map(country_territory_map))

# If a country has NO territory info in the entire dataset, fill with 'Unknown'
df['TERRITORY'] = df['TERRITORY'].fillna('Unknown')


# 2. Handle ADDRESSLINE2: Merge into ADDRESSLINE1
mask = df['ADDRESSLINE2'].notnull() # Identify rows where AddressLine2 is not null
df.loc[mask,'ADDRESSLINE1'] = df.loc[mask, 'ADDRESSLINE1'] + ', ' + df.loc[mask, 'ADDRESSLINE2']# Concatenate: "Address1, Address2"

# Drop the now redundant column
df.drop('ADDRESSLINE2', axis=1, inplace=True)


# 3. Standard Imputation for other fields
df['POSTALCODE'] = df['POSTALCODE'].fillna('Unknown')
df['STATE'] = df['STATE'].fillna('Not Applicable')
print("Data cleaned successfully.")
```

```python
# Min, avg, and max for Quantity Ordered and Price Each.
# Group by Product Line and calculate statistics
stats = df.groupby('PRODUCTLINE')[['QUANTITYORDERED', 'PRICEEACH']].agg(['min', 'mean', 'max'])
print(stats)
```

```python
# Filter & Group for 2004
# Ensure ORDERDATE is datetime
df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'])
df['YEAR'] = df['ORDERDATE'].dt.year

# Filter dataset for the year 2004
sales_2004 = df[df['YEAR'] == 2004]

# Group by Product Line and Country, then sum Sales
grouped_sales_2004 = sales_2004.groupby(['PRODUCTLINE', 'COUNTRY'])['SALES'].sum().reset_index()

# Display Result
print("Total Sales per Product Line and Country (2004 Only):")
print(grouped_sales_2004)
```

```python
# Sales Volume by Deal Size
deal_size_sales = df.groupby('DEALSIZE')['SALES'].sum()
print("Sales Volume by Deal Size")
print(deal_size_sales)
print("") # Creates a blank line

# Top 10 Customers
top_customers = df.groupby('CUSTOMERNAME')['SALES'].sum().nlargest(10)
print("Top 10 Customers")
print(top_customers)

# Visualize Sales Trends
# Group by date
daily_sales = df.groupby('ORDERDATE')['SALES'].sum()

# Plot
plt.figure(figsize=(12, 6))
daily_sales.plot(kind='line', color='blue')
plt.title('Sales Trends Over Time')
plt.ylabel('Total Sales ($)')
plt.xlabel('Date')
plt.grid(True)
plt.show()
```

```python
# NumPy Statistical Analysis (Outliers & Comparison)
# We use NumPy to calculate Z-scores for outlier detection and compare sales between regions.

sales_data = df['SALES'].values

# 1. Outlier Detection (Z-Score > 3)
mean_sales = np.mean(sales_data)
std_sales = np.std(sales_data)
z_scores = (sales_data - mean_sales) / std_sales

outliers = sales_data[np.abs(z_scores) > 3]

print(f"Mean Sales: ${mean_sales:.2f}")
print(f"Number of Outliers (Z > 3): {len(outliers)}")
print(f"Max Outlier Transaction: ${np.max(outliers):.2f}")

# 2. Statistical Comparison: USA vs. Rest of World
usa_sales = df[df['COUNTRY'] == 'USA']['SALES'].values
row_sales = df[df['COUNTRY'] != 'USA']['SALES'].values

print(f"Avg Sales (USA): ${np.mean(usa_sales):.2f}")
print(f"Avg Sales (Rest of World): ${np.mean(row_sales):.2f}")
```

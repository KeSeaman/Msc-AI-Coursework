# AI Governance and Ethics: Frameworks and Challenges

## 1. Introduction

The rapid proliferation of Artificial Intelligence (AI), particularly General Purpose AI (GPAI) and foundation models, has transformed the global socio-economic landscape. Once confined to speculative technical research, AI is now deeply embedded in critical infrastructures, public administration, financial markets, and the fabric of daily life. This ubiquity has brought the governance and ethical deployment of these systems to the forefront of global policy debates. Because AI systems shape decision-making in vital areas like credit allocation, healthcare diagnostics, and criminal justice, their design and deployment are no longer purely technical choices but profound political and ethical exercises.

However, contemporary discourse on AI governance is heavily characterized by a geographical and institutional imbalance. Regulatory models, risk taxonomies, and ethical guidelines are primarily drafted and championed by developed economies, most notably the European Union (EU) and the United States (US). These frameworks naturally reflect the regulatory capacities, market structures, and socio-economic priorities of the Global North. For developing nations, particularly in the Global South, this asymmetry presents a significant challenge: governance architectures designed for capital-rich, highly institutionalized environments often fail to align with the developmental imperatives, severe resource constraints, and democratic aspirations of emerging economies.

This term paper examines the multi-faceted landscape of AI governance and ethics, drawing key insights from the *Ethics and Governance of AI Synthesis Report* (Kanwar, 2026), published by the ICRIER Prosus Centre for Internet and Digital Economy (IPCIDE). The report synthesizes discussions from a monthly webinar series convened from October 2025 to January 2026, which featured dialogues between global experts and Indian policy discussants. By examining the tensions between global standards and local realities, this paper explores how countries like India and the wider BRICS block can navigate the complexities of regulating fluid value chains, embedding sustainability across the AI lifecycle, protecting digital sovereignty, and operationalizing equity and accountability in automated systems.

## 2. Problem Statement

The central conflict in contemporary AI governance lies in the structural misalignment between the globalized, highly concentrated supply of AI technologies and the localized, diverse contexts of their deployment. This conflict manifests across four distinct systemic crises:

### Asymmetries in the AI Value Chain
The GPAI value chain is non-linear and dominated by a handful of upstream foundation model providers based primarily in the Global North. Downstream developers and deployers in the Global South face intense information and bargaining asymmetries. While downstream actors bear the immediate compliance, operational, and reputational risks of integrating these models into local applications, they lack the access to verify the safety, training criteria, and embedded biases of the underlying upstream models. Upstream providers frequently shield themselves from liability using contract terms, leaving downstream developers exposed to legal liabilities for automated outcomes they cannot control.

### Environmental and Material Exploitation
The computational intensity required to train and run massive foundation models translates to a significant physical and ecological footprint. Data centers demand massive electrical power grids and millions of liters of water for cooling. This material reality places a disproportionate ecological strain on developing nations, where energy security is already a developmental challenge and water scarcity is an existential threat. The resources required to power foreign AI systems are extracted from local ecosystems, effectively exporting the environmental costs of AI to the Global South while concentrating its profits and control in the Global North.

### Digital and AI Sovereignty Deficit
Relying exclusively on proprietary, foreign-developed AI models creates a dependency that compromises the self-determination of developing states. These foreign models naturally embed the cultural values, language models, and socio-political norms of their creators. Without indigenous technological alternatives and local compute infrastructure, states risk becoming mere consumers of systems that are poorly suited to their local realities, developmental priorities, and demographic diversity. This undermines their digital sovereignty and weakens their capacity to regulate public interests.

### Technocratic Public Engagement Gaps
AI decision-making remains highly technocratic and insulated from public accountability. There is a structural disconnect between AI researchers, policy makers, and the public. Under a "deficit model" of public engagement, policy makers merely seek to educate the public about AI after the fact, rather than involving them in the upstream design and governance of the technologies that will govern their lives. This leaves the public as passive consumers rather than active designers, creating a democratic deficit that undermines trust and increases the risk of systemic algorithmic harms.

## 3. Overview of AI Governance

AI governance refers to the collective set of rules, practices, institutions, and technical standards designed to steer the development, deployment, and operational lifecycle of artificial intelligence systems. It aims to align AI systems with public interest values (such as safety, fairness, transparency, and accountability) while preventing harms. To be effective, AI governance must span three distinct levels:

```mermaid
graph TD
    A[AI Governance Stack] --> B[Data Governance]
    A --> C[Algorithmic Governance]
    A --> D[Infrastructure Governance]
    B --> B1["Personal Privacy & Consent"]
    B --> B2["Non-Personal Data Commons"]
    B --> B3["Data Localization Frameworks"]
    C --> C1["Model Alignment & Safety"]
    C --> C2["Bias Audits & Evaluations"]
    C --> C3["Explainability & Transparency"]
    D --> D1["Sovereign Compute & Cloud"]
    D --> D2["Energy & Water Mandates"]
    D --> D3["Connectivity & Broadband Access"]
```

Historically, digital governance focused primarily on **data governance**, regulating data privacy, cross-border data flows, and localization. However, the rise of foundational AI has necessitated a shift toward **AI sovereignty**. AI sovereignty extends digital sovereignty by asserting a nation's capacity to domestically understand, develop, and regulate AI systems. As noted in the ICRIER webinar discussions, risk-based regulations alone are insufficient. A state cannot achieve true sovereignty simply by banning harmful systems. True AI sovereignty requires establishing the foundational enablers to build domestic technological alternatives that reflect local cultural values and serve local needs. It represents a move from passive consumerism to active self-determination in the global technology stack.

## 4. Current Frameworks

### Global Regulatory Landscapes & The 2026 Milestones
The global regulatory environment is currently undergoing a critical transition from theoretical compliance to active enforcement, marked by major legislative deadlines in mid-2026. The **EU Artificial Intelligence Act (2024)** remains the primary benchmark. Its most significant milestone approaches on **August 2, 2026**, when the majority of the Act's rules enter into full force. This includes the stringent **Article 50 transparency obligations**, which mandate clear disclosures for human-interacting AI, deepfake labels, and machine-readable watermarking for synthetic audio, video, or text. Legacy systems already on the market are granted a brief grace period until **December 2, 2026**, to implement these watermarks.

To manage practical implementation bottlenecks, the EU recently adopted the **Digital Omnibus on AI** (with final political agreement reached in **May 2026**). This omnibus bill provides necessary breathing room by postponing the mandate for Member States to set up regulatory sandboxes to **August 2, 2027**, and extending the deadline for high-risk AI embedded in regulated products to **August 2, 2028**.

Simultaneously, the United States has seen a surge in state-level statutory frameworks rather than federal legislation. Key developments in 2026 include the enactment of the **Connecticut AI Responsibility and Transparency (CART) Act**, which establishes a modular safety gate for frontier models, and Colorado's revised AI Act. In California, the implementation of the **Delete Request and Opt-Out Platform (DROP)** starting **August 1, 2026**, forces data brokers to systematically process deletion requests, severely limiting the data pools available for unauthorized corporate model fine-tuning.

### The Upstream-Downstream Allocation Gap
A key critique of current frameworks is the asymmetric distribution of obligations across the AI value chain. The Act places heavy compliance burdens on upstream GPAI developers. However, in practice, upstream providers use end-user license agreements (EULAs) and liability-shifting clauses to pass operational and compliance risks downstream. Consequently, downstream developers (often small and medium enterprises in emerging markets) face a double bind: they are legally responsible for deployment outcomes, but lack the technical information, access keys, and bargaining power to audit the upstream API models they rely on. This is exacerbated by a "cooperation deficit," where upstream providers refuse to share critical validation data, citing proprietary trade secrets.

### The Role of Private Sector Governance & Antitrust Interventions
In the absence of harmonized global public law, the private sector has historically relied on voluntary commitments (like the Frontier Model Forum) and internal ethics boards, which webinar insights characterize as "ethics washing" designed to pre-empt binding legislation.

However, the year 2026 has witnessed a major shift as public antitrust and competition regulators begin to intervene in private data monopolies. In **June 2026**, the U.K. Competition and Markets Authority (CMA) issued a landmark directive ordering Google to provide digital publishers with granular opt-out controls, preventing the search giant from scraping local publisher content for AI-generated search summaries without consent. This highlights a growing regulatory consensus that private corporate governance is insufficient to prevent anticompetitive value extraction in the AI ecosystem.

## 5. Ethical Considerations in AI

The ethical deployment of AI requires translating abstract principles into concrete institutional and technical safeguards. The ICRIER Synthesis Report highlights four key ethical considerations, updated with landmark 2026 legal developments:

### Bias and Fairness
AI models are trained on datasets heavily concentrated in the Global North. This leads to "representational bias," where models incorporate Western cultural norms, idioms, and classifications while failing to comprehend local dialects, social structures, and contextual nuances of the Global South. For example, automated resume screening or medical diagnosis models trained on demographic data from developed nations exhibit high error rates when deployed in rural clinics or diverse labor markets in emerging economies. Fairness must therefore be understood not merely as mathematical statistical parity, but as a contextual commitment to historical and structural justice, preventing the marginalization of vulnerable groups.

The boundary between human authorship and machine generation was further solidified in **March 2026**, when the U.S. Supreme Court declined to review *Thaler v. Perlmutter*, establishing a binding precedent that AI-generated works cannot be granted copyright protection, reinforcing the ethical baseline that authorship remains an exclusively human domain.

### Privacy and Data Protection
The training of foundation models relies on massive web-scraping pipelines that harvest personal, creative, and intellectual data without consent or compensation. In 2026, the legal backlash against this has culminated in over **80 active copyright and privacy infringement lawsuits** globally. Authors have filed class-action suits against Adobe, and prominent content creators have sued Meta, alleging that model training circumvented Technical Protection Measures (TPMs) and systematically scraped protected creative works. Data governance frameworks must establish robust boundaries against predatory data extraction. This includes enforcing data localization mandates, defining clear consent mechanisms for web-crawlers, and creating public "data commons" where non-personal data is treated as a collective national resource rather than private corporate property.

### Transparency and Explainability
Deep learning architectures are inherently "black boxes," where billions of parameter weights make it nearly impossible to trace the exact rationale behind a specific algorithmic output. This lack of transparency is highly problematic when AI is used to allocate public benefits, determine creditworthiness, or assist in judicial decisions. Governance frameworks must mandate **Explainable AI (XAI)** standards, requiring developers to provide human-interpretable explanations of model logic. Furthermore, intermediate transparency (such as model cards, system cards, and datasheets for datasets) must be legally required to address the information asymmetries between developers and users.

### Accountability and Responsibility
When an AI system causes harm, identifying the liable party is legally complex. Current legal frameworks must establish clear joint-and-several liability regimes across the AI value chain. Upstream model developers must be held accountable for structural failures, security vulnerabilities, and embedded biases in their base models, while downstream deployers remain responsible for application-specific contextual failures. This requires establishing statutory cooperation duties, forcing upstream developers to provide downstream integrators with detailed technical documentation and auditing capabilities.

Furthermore, a critical 2026 legal development has emerged regarding **confidentiality and attorney-client privilege**. Federal courts in jurisdictions like the Southern District of New York (SDNY) and Colorado ruled in early 2026 that uploading proprietary or sensitive data to public generative AI platforms waives legal privilege, classifying the AI model host as an unauthorized third-party recipient. This has forced organizations to implement strict technical guardrails to prevent data leaks.

## 6. Future Considerations

As AI continues to evolve, future governance frameworks must transition from static risk-management models to active, adaptive, and holistic enablers. Three key structures will define this future:

### 1. Operationalising the SCAIS Framework
To address the environmental and social sustainability challenges of AI, future policies should adopt the **Sustainability Criteria and Indicators for Artificial Intelligence Systems (SCAIS)** framework. Rather than evaluating AI purely on efficiency or accuracy, the SCAIS framework measures performance across four integrated dimensions:

| SCAIS Dimension | Key Governance Criteria | Target Indicators |
| :--- | :--- | :--- |
| **Social** | Equity of access, cultural diversity, labor rights, bias mitigation | Gender representation in training data, fair wages for data labelers, local language support |
| **Ecological** | Environmental resource preservation, carbon reduction, planetary boundaries | Data center power usage effectiveness (PUE), water consumption per query, hardware recycling rate |
| **Economic** | Common good orientation, market diversity, anti-monopoly, open access | Open-source model availability, compute access subsidies for startups, local SME participation |
| **Governance** | Transparency, systemic accountability, regulatory adaptability | Model card publishing rates, audit trail completeness, incident reporting speed |

### 2. Securing the AI Sovereignty Stack (KASE Framework)
To prevent technological dependency, countries in the Global South should invest in the **Key AI Sovereignty Enablers (KASE)** framework. AI sovereignty cannot be achieved through regulation alone; it requires building and maintaining an independent technical stack:
- **Data Governance**: Creating localized data repositories and collective data rights to leverage data as a national resource.
- **Algorithmic R&D**: Investing in public research to develop models optimized for low-resource environments and local languages.
- **Computational Capacity**: Building sovereign compute centers and high-performance computing (HPC) clusters powered by renewable energy.
- **Meaningful Connectivity**: Ensuring universal, affordable broadband access to democratize AI benefits.
- **Sustainable Power**: Aligning AI infrastructure with green energy transitions to avoid carbon-intensive data processing.
- **Digital Literacy**: Training a domestic workforce to produce, rather than merely consume, AI systems.

### 3. AI as Digital Public Infrastructure (DPI)
Rather than leaving the AI market to the exclusive control of monopolistic private firms, future governance should treat foundational AI models and computational power as public utilities. By building sovereign, open-source AI models and compute pools, similar to India's Aadhaar, UPI, and ONDC frameworks, governments can democratize access, ensure strict compliance with public interest ethics, and foster local innovation. This shifts the focus from private value extraction to public value creation.

## 7. Conclusions and Recommendation

### Conclusions
AI governance stands at a critical juncture. The current dominance of Global North regulatory models has created an asymmetry that exposes downstream developers and developing nations to disproportionate risks, dependencies, and environmental costs. Effective AI governance cannot rely solely on self-regulation by private monopolists or static, risk-based frameworks. True ethical alignment requires a holistic approach that integrates value chain accountability, ecological sustainability, and the cultivation of sovereign technological capabilities. Platforms like the 2026 BRICS presidency and the AI Impact Summit provide timely opportunities for the Global South to propose alternative, non-aligned visions for AI governance.

### Recommendations for Policy Makers
1. **Enforce Value Chain Accountability**: Legislate statutory information-sharing and cooperation duties that force upstream foundation model developers to provide transparency and share liability with downstream integrators.
2. **Mandate Environmental Disclosures**: Require AI developers and data center operators to publicly disclose their energy consumption, water usage, carbon footprints, and supply chain mineral sourcing (aligning with the SCAIS ecological criteria).
3. **Invest in Sovereign Compute and Open-Source Models**: Establish national and multilateral compute pools powered by renewable energy, and fund open-source, localized foundation models tailored to low-resource languages and domestic developmental needs.
4. **Reconstruct Public Engagement**: Abandon the "deficit model" of public education and implement participatory mechanisms (such as citizen assemblies, multi-stakeholder advisory panels, and community consultation loops) to shape AI deployment boundaries.
5. **Harness BRICS and Global South Solidarity**: Leverage platforms like the BRICS presidency and the AI Impact Summit to establish alternative, non-aligned standards for AI sovereignty, fostering mutual compute-sharing agreements and localized data commons.
